## D2. Pengenalan Python untuk membaca dan menampilkan image dengan  pengenalan macam-macam Library (Numpy, pandas, cv2, skimage, PIL dan matplotlib)



```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from PIL import Image
from matplotlib import pyplot as plt
```

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016008/full/800,/0/default.jpg"]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
    image = io.imread(url)                                #read image
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5)      #resize image to half size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)       #convert color to RGB
    final_frame = cv.hconcat((image, image_2))           #concatenate image
    cv2_imshow(final_frame)                              #show image
    print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
    image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
# Create vertical lines and diagonal lines in the output image

import numpy as np

# Ukuran gambar
tinggi = image_2.shape[0]
lebar = image_2.shape[1]


img_vertikal = image_2.copy()
for x in range(tinggi):
    img_vertikal[x, int(lebar / 2)] = [255, 255, 255]


img_silang = image_2.copy()
for y in range(tinggi):
    x1 = int(y * (lebar - 1) / (tinggi - 1))
    x2 = lebar - 1 - x1
    img_silang[y, x1] = [255, 255, 255]
    img_silang[y, x2] = [255, 255, 255]


final_frame = cv.hconcat((image_2, img_vertikal, img_silang))
cv2_imshow(final_frame)
```

```python
# Draw the horizontal white lines in the middle of image with certain length
import cv2 as cv
from google.colab.patches import cv2_imshow

img_garis = image_2.copy()
tinggi, lebar = img_garis.shape[:2]


panjang_garis = 200

y_tengah = tinggi // 2
x_start = (lebar - panjang_garis) // 2
x_end = x_start + panjang_garis


for x in range(x_start, x_end):
    img_garis[y_tengah, x] = [255, 255, 255]


final_frame = cv.hconcat((image_2, img_garis))
cv2_imshow(final_frame)
```

```python
# Draw a box from a collection of white pixel in certain area in the image
img_box = image_2.copy()
tinggi, lebar = img_box.shape[:2]


panjang_kotak = 100
tinggi_kotak = 80


x_start = (lebar - panjang_kotak) // 2
x_end = x_start + panjang_kotak

y_start = (tinggi - tinggi_kotak) // 2
y_end = y_start + tinggi_kotak


img_box[y_start:y_end, x_start:x_end] = [255, 255, 255]



final_frame = cv.hconcat((image_2, img_box))
cv2_imshow(final_frame)
```

### 1. `figsize` vs. Ukuran Pixel Gambar

`figsize` mengatur ukuran *tampilan* plot di Matplotlib, tidak mengubah resolusi pixel sebenarnya dari gambar itu sendiri. Kita akan buktikan dengan menampilkan `img.shape` dan kemudian menampilkan gambar dengan `figsize` yang berbeda.

```python
# Load the image again to ensure we're starting fresh for this example
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/drive/MyDrive/Semester 5/PCVK/KTM.jpeg")
img2 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


tinggi_asli, lebar_asli, _ = img2.shape
print(f"Ukuran pixel gambar asli (img2): {lebar_asli} x {tinggi_asli}")

# Tampilkan dengan figsize default
plt.figure(figsize=(6, 6))
plt.imshow(img2)
plt.title("Gambar dengan figsize default (6,6)")
plt.show()

# Tampilkan dengan figsize yang lebih besar
plt.figure(figsize=(12, 12))
plt.imshow(img2)
plt.title("Gambar dengan figsize (12,12)")
plt.show()

# Periksa lagi ukuran pixel gambar
tinggi_setelah_figsize, lebar_setelah_figsize, _ = img2.shape
print(f"Ukuran pixel gambar setelah ditampilkan dengan figsize berbeda: {lebar_setelah_figsize} x {tinggi_setelah_figsize}")

```

### 5. Buat 5 Kotak Berbagai Ukuran dan Warna yang Berbeda (Acak) dalam Satu Image!

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image again to ensure we're starting fresh for this example
img = cv2.imread("/content/drive/MyDrive/Semester 5/PCVK/KTM.jpeg")
img2 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Buat salinan gambar untuk digambar
image_with_boxes = img2.copy()

tinggi, lebar, _ = image_with_boxes.shape

for _ in range(5):
    # Ukuran kotak acak
    box_width = np.random.randint(20, 150)
    box_height = np.random.randint(20, 150)

    # Posisi acak (pastikan kotak tidak keluar dari gambar)
    x1 = np.random.randint(0, lebar - box_width)
    y1 = np.random.randint(0, tinggi - box_height)
    x2 = x1 + box_width
    y2 = y1 + box_height

    # Warna acak (RGB)
    color = (np.random.randint(0, 256), np.random.randint(0, 256), np.random.randint(0, 256))

    # Ketebalan garis acak (atau -1 untuk mengisi)
    thickness = np.random.randint(1, 10)
    # uncomment baris di bawah untuk kotak terisi penuh
    # thickness = -1

    cv2.rectangle(image_with_boxes, (x1, y1), (x2, y2), color, thickness)

plt.figure(figsize=(10, 10))
plt.imshow(image_with_boxes)
plt.title('Image dengan 5 Kotak Acak')
plt.show()
```
