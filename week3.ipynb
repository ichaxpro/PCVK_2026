```python
# transformasi logarithmic brightness

import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Mengubah tingkat kecerahan citra dengan Transformasi Log')
print('------------------------------------------------------')

try:
    brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Error, bukan angka. Menggunakan nilai default 50')
    brightness = 50

original = cv.imread('/content/drive/MyDrive/Semester 5/PCVK/FOTO_1.jpg')

img_float = original.astype(np.float32)


# c = brightness / log(1 + r_max), atau menggunakan konstanta langsung yang diskalakan
# Formula umum: s = c * log(1 + r)
c = brightness / np.log(1 + np.max(img_float))
log_image = c * np.log(1 + img_float)

log_image = np.clip(log_image, 0, 255).astype(np.uint8)

final_frame = cv.hconcat([original, log_image])

cv2_imshow(final_frame)
```

```python
# Grayscale

import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Transformasi Grayscale (Averaging, Lightness, Luminance)')
print('---------------------------------------------------------')

original = cv.imread('/content/drive/MyDrive/Semester 5/PCVK/paprika.jpg')

# Pisahkan channel warna B, G, R dari citra asli
# OpenCV membaca citra dalam urutan BGR (Blue = index 0, Green = index 1, Red = index 2)
b, g, r = cv.split(original)
r_f = r.astype(np.float32)
g_f = g.astype(np.float32)
b_f = b.astype(np.float32)


# a. Metode Averaging
# Formula: Gray = (R + G + B) / 3
gray_averaging = ((r_f + g_f + b_f) / 3).astype(np.uint8)

# b. Metode Lightness
# Formula: Gray = (max(R, G, B) + min(R, G, B)) /
max_rgb = np.maximum(np.maximum(r_f, g_f), b_f)
min_rgb = np.minimum(np.minimum(r_f, g_f), b_f)
gray_lightness = ((max_rgb + min_rgb) / 2).astype(np.uint8)

# c. Metode Luminance
# Formula: Gray = 0.299*R + 0.587*G + 0.114*B
gray_luminance = (0.299 * r_f + 0.587 * g_f + 0.114 * b_f).astype(np.uint8)

img_avg_bgr = cv.cvtColor(gray_averaging, cv.COLOR_GRAY2BGR)
img_light_bgr = cv.cvtColor(gray_lightness, cv.COLOR_GRAY2BGR)
img_lum_bgr = cv.cvtColor(gray_luminance, cv.COLOR_GRAY2BGR)

final_frame = cv.hconcat([original, img_avg_bgr, img_light_bgr, img_lum_bgr])

cv2_imshow(final_frame)
```

```python
# memanipulasi warna

import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Menampilkan Warna Tertentu (Merah) dan Lainnya Grayscale')
print('---------------------------------------------------------')

original = cv.imread('/content/drive/MyDrive/Semester 5/PCVK/paprika.jpg')

hsv = cv.cvtColor(original, cv.COLOR_BGR2HSV)


# Karena warna merah berada di dua rentang ujung nilai Hue (0-10 dan 170-180):
lower_red1 = np.array([0, 100, 100])
upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([170, 100, 100])
upper_red2 = np.array([180, 255, 255])

mask1 = cv.inRange(hsv, lower_red1, upper_red1)
mask2 = cv.inRange(hsv, lower_red2, upper_red2)
mask_red = cv.bitwise_or(mask1, mask2)

#Buat  Grayscale,
gray = cv.cvtColor(original, cv.COLOR_BGR2GRAY)
gray_bgr = cv.cvtColor(gray, cv.COLOR_GRAY2BGR)

# Ambil bagian asli yang berwarna merah
red_extracted = cv.bitwise_and(original, original, mask=mask_red)

mask_inv = cv.bitwise_not(mask_red)
background_gray = cv.bitwise_and(gray_bgr, gray_bgr, mask=mask_inv)

final_image = cv.add(red_extracted, background_gray)

final_frame = cv.hconcat([original, final_image])

cv2_imshow(final_frame)
```

```python
# Gamma Correction

import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print(' Gamma Correction pada citra ')
print('----------------------------------')

try:
    gamma = float(input('Masukkan nilai Gamma (contoh: 2.2 atau 0.5): '))
except ValueError:
    print('Error, bukan angka. Menggunakan nilai default 1.0')
    gamma = 1.0

original = cv.imread('/content/drive/MyDrive/Semester 5/PCVK/FOTO_1.jpg')

# Rumus: O = 255 * (I / 255) ^ (1 / gamma)
invGamma = 1.0 / gamma
table = np.array([((i / 255.0) ** invGamma) * 255 for i in np.arange(0, 256)]).astype("uint8")

gamma_image = cv.LUT(original, table)

final_frame = cv.hconcat([original, gamma_image])

cv2_imshow(final_frame)
```

```python
# Image Depth

import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Simulasi Image Depth (Kuantisasi Citra)')
print('---------------------------------------')

try:
    bit_depth = int(input('Masukkan bit depth tujuan (1-8): '))
except ValueError:
    print('Error, bukan angka. Menggunakan nilai default 3')
    bit_depth = 3

# Menghitung jumlah level warna berdasarkan bit depth tujuan
# Contoh: jika 3 bit, maka 2^3 = 8 level (0 sampai 7)
num_levels = 2 ** bit_depth
level = 255 / (num_levels - 1)

original = cv.imread('/content/drive/MyDrive/Semester 5/PCVK/paprika.jpg', cv.IMREAD_GRAYSCALE)

# Proses kuantisasi citra (mengurangi kedalaman bit / image depth)
# Rumus: pembagian dengan level, dibulatkan, lalu dikali kembali dengan level
depth_image = np.round(original / level) * level
depth_image = depth_image.astype(np.uint8)

print(f"Bit depth awal  : 8 bit")
print(f"Bit depth tujuan: {bit_depth} bit")
print(f"Jumlah level    : {num_levels}")

final_frame = cv.hconcat([original, depth_image])

cv2_imshow(final_frame)
```

```python
import cv2 as cv
import numpy as np
import glob
from google.colab.patches import cv2_imshow

print('Simulasi Average Denoising & Perhitungan PSNR')
print('---------------------------------------------')

original_path = '/content/drive/MyDrive/Images/galaxy.jpg'
original = cv.imread(original_path)

cv_img = []
for img in glob.glob('/content/drive/MyDrive/Images/noises/*.jpg'):
    n = cv.imread(img)
    if n is not None:
        cv_img.append(n)

print(f"Total citra noise yang berhasil dimuat: {len(cv_img)} citra\n")

#Fungsi untuk melakukan Average Denoising & Hitung PSNR
def run_average_denoising(num_images):
    from google.colab.patches import cv2_imshow

    if len(cv_img) < num_images:
        print(f"Peringatan: Jumlah citra tersedia ({len(cv_img)}) kurang dari {num_images}!")
        return

    subset = cv_img[:num_images]
    accumulated = np.zeros_like(subset[0], dtype=np.float32)

    for img in subset:
        accumulated += img.astype(np.float32)

    averaged_img = (accumulated / num_images).astype(np.uint8)
    psnr_value = cv.PSNR(original, averaged_img)

    print(f"Jumlah Citra di-Average : {num_images}")
    print(f"Nilai PSNR (dB)        : {psnr_value:.4f}")
    print("-" * 45)

    final_frame = cv.hconcat([original, averaged_img])
    cv2_imshow(final_frame)
    return averaged_img

# --- UJI COBA SESUAI TABEL ---
print("--- Uji Coba 10 Citra ---")
run_average_denoising(10)

print("\n--- Uji Coba 20 Citra ---")
run_average_denoising(20)

print("\n--- Uji Coba 40 Citra ---")
run_average_denoising(40)

print("\n--- Uji Coba 80 Citra ---")
run_average_denoising(80)

print("\n--- Uji Coba 100 Citra ---")
run_average_denoising(100)
```

Based on the simulation results, yes, as the number of averaged images increases (from 10 up to 100 images), the PSNR (Peak Signal-to-Noise Ratio) value also increases progressively from 19.7249 dB to 19.9220 dB. This demonstrates that averaging more noisy frames effectively reduces random Gaussian noise, as independent noise components tend to cancel each other out. Consequently, the quality of the reconstructed image improves and grows closer to the original clean image as the sample size becomes larger.

```python
# Image Masking


import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Praktikum Image Masking dengan Berbagai Operator Logika')
print('---------------------------------------------------------')

img_path = '/content/drive/MyDrive/Semester 5/PCVK/couple.jpg'
original = cv.imread(img_path)

h, w = original.shape[:2]

mask = np.zeros((h, w), dtype=np.uint8)
cv.circle(mask, (int(w * 0.35), int(h * 0.3)), int(h * 0.25), 255, -1) # Lingkaran kiri
cv.circle(mask, (int(w * 0.65), int(h * 0.3)), int(h * 0.25), 255, -1) # Lingkaran kanan

mask_bgr = cv.cvtColor(mask, cv.COLOR_GRAY2BGR)

print("Memproses operator logika...")

# 1. OPERATOR NOT (Komplemen pada Mask)
not_mask = cv.bitwise_not(mask)
res_not = cv.bitwise_and(original, original, mask=not_mask)

# 2. OPERATOR OR
res_or = cv.bitwise_or(original, mask_bgr)

# 3. OPERATOR AND (Standard Masking)
res_and = cv.bitwise_and(original, original, mask=mask)

# 4. OPERATOR NAND (Not AND)
and_masked = cv.bitwise_and(original, original, mask=mask)
res_nand = cv.bitwise_not(and_masked)

# 5. OPERATOR XOR
res_xor = cv.bitwise_xor(original, mask_bgr)

# --- MENAMPILKAN HASIL ---
print("\n--- Menampilkan Hasil Operator ---")
print("1. Mask Utama:")
cv2_imshow(mask_bgr)

print("2. Hasil Operator AND (Masking Standar):")
cv2_imshow(res_and)

print("3. Hasil Operator NOT:")
cv2_imshow(res_not)

print("4. Hasil Operator OR:")
cv2_imshow(res_or)

print("5. Hasil Operator XOR:")
cv2_imshow(res_xor)

print("6. Hasil Operator NAND:")
cv2_imshow(res_nand)
```

```python
# 10. Using Linear Brightness

import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow


print('Mengubah tingkat kecerahan citra')
print('----------------------------------')
try:
  brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
  print('Error, not a number')

original = cv.imread('/content/drive/MyDrive/Semester 5/PCVK/malam.jpg')

brightness_image = cv.convertScaleAbs(original, beta=brightness)

final_frame = cv.hconcat([original, brightness_image])

cv2_imshow(final_frame)
```

- Reason: It is the simplest method because it only requires adding a constant value to every pixel to brighten up a dark night image instantly.
- Risk: It easily causes overexposure (saturation), making bright areas or lights completely white (washed out) and increasing background noise significantly.

```python
# 11. Perbaiki kualitas pada citra

import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Enhancing Crayfish Image (Contrast & Gamma)')
print('-------------------------------------------')

original = cv.imread('/content/drive/MyDrive/Images/crayfish.jpg')

# Apply Contrast Stretching / Linear Adjustment (Alpha = 1.5, Beta = 30)
alpha = 1.5
beta = 30
adjusted_contrast = cv.convertScaleAbs(original, alpha=alpha, beta=beta)

# Apply Gamma Correction (Gamma = 0.5 for brightening shadows)
gamma = 0.5
invGamma = 1.0 / gamma
table = np.array([((i / 255.0) ** invGamma) * 255 for i in np.arange(0, 256)]).astype("uint8")
final_enhanced = cv.LUT(adjusted_contrast, table)

# Display original and enhanced result side by side
final_frame = cv.hconcat([original, final_enhanced])
cv2_imshow(final_frame)
```

The crayfish.jpg image is very dark and has low contrast because it was taken underwater. Combining Contrast Stretching expands the pixel intensity range, while Gamma Correction brightens the dark hidden details without completely washing out the background.
